1. ¿Cuál es el continente con la mayor cantidad de países en el conjunto de datos? Proporciona el porcentaje de países en ese continente.


```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None) # para poder visualizar todas las columnas de los DataFrames
```

```python
df = pd.read_csv("world_data_full_apply_nulos.csv", index_col=0)

df.head(5)
```

```python
df.columns
```

```python
df_continentes = df['continent'].value_counts()

categorias = df_continentes.index
```

```python
categorias
```

```python
# Crear el gráfico de pastel
plt.pie(df_continentes, labels=categorias, autopct='%1.2f%%',colors=['c', 'lightblue', 'skyblue'])

# Título
plt.title("Distribución de países por continente")

# Mostrar la gráfica
plt.show()
```

```python
plt.bar(x=categorias, height=df_continentes.values, color='c')
plt.xticks(rotation=90)
plt.xlabel("Continente")
plt.ylabel("Nº países")
```

2. ¿Cuáles son los 10 países con la mayor área forestal?


```python
df_area_forestal = df[['country','forested_area']].sort_values(by='forested_area', ascending=False).head(10)

df_area_forestal
```

```python
sns.barplot(x = "country", y = "forested_area", data = df_area_forestal, color="green")
plt.ylabel("area forestal")
plt.xlabel("pais")
plt.xticks(rotation=90);
```

3. ¿Cómo se distribuye la densidad de población entre los países?


```python
df.plot(kind='hist', y='density', bins=20, figsize=(10, 6), color='violet', edgecolor='black')

# Añadir títulos y etiquetas
plt.title("Distribución densidad población")
plt.xlabel("Densidad")
plt.ylabel("Paises")

# Mostrar el gráfico
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(y='density', data=df)
plt.title('Distribución de densidad por paises')
plt.xticks(rotation=45)  # Gira etiquetas para que se vean
plt.show()
```

4. ¿Existe una relación entre el PIB y la tasa de desempleo de los países?


```python
sns.regplot(x=df['unemployment_rate'], y=df['gdp'], color="red")

plt.title("Relación PIB tasa desempleo")
plt.xlabel("Tasa desempleo")
plt.ylabel("PIB")

plt.show()
```

5. ¿Cómo se relacionan la tasa de mortalidad infantil y la esperanza de vida?


```python
df.columns
```

```python
sns.regplot(x=df['infant_mortality'], y=df['life_expectancy'], color="orange")
plt.title('Gráfico de Dispersión: mortalidad infantil vs esperanza de vida')
plt.xlabel('Mortalidad infantil')
plt.ylabel('Esperanza de vida')
plt.show()
```


> Requisitos que deben cumplir las gráficas (en caso de usar alguna de ellas):
>
> * Todas deben llevar título.
> * Los nombres de los ejes deben estar en castellano.
> * En los *scatterplot* los puntos deben tener un tamaño de 2.
> * En los *scatterplot* los puntos deben tener un color diferente al azul.
> * En los *barplot* las barras siempre tienen que estar ordenadas de mayor a menor.
> * En los *barplot* y *countplot* las barras tienen que tener un color diferente a los que vienen por defecto.
> * En los *pieplot* siempre tenemos que añadir la etiqueta de a que % corresponde cada quesito.
