### **Ejercicios visualización**

A lo largo de este ejercicio plantearemos una serie de preguntas que tendrás que contestar usando la visualización (podréis usar tanto Seaborn como Matplotlib).

Requisitos que deben cumplir las gráficas (en caso de usar alguna de ellas):

* Todas deben llevar título.
* Los nombres de los ejes deben estar en castellano.
* En los *scatterplot* los puntos deben tener un tamaño de 2.
* En los *scatterplot* los puntos deben tener un color diferente al azul.
* En los *barplot* las barras siempre tienen que estar ordenadas de mayor a menor.
* En los *barplot* y *countplot* las barras tienen que tener un color diferente a los que vienen por defecto.
* En los *pieplot* siempre tenemos que añadir la etiqueta de a que % corresponde cada quesito.


```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None) # para poder visualizar todas las columnas de los DataFrames
```

```python
df = pd.read_csv("world_data_full_apply_nulos.csv", index_col = 0)
```

```python
df.head(5)
```


1. ¿Cómo se distribuye la tasa de natalidad en el conjunto de datos? ¿Hay algún patrón evidente?


```python
df.columns
```

```python
df_natalidad = df.sort_values(by= "fertility_rate", ascending = True) 
```

```python
plt.hist("fertility_rate", data = df_natalidad, bins = 10, color = "red", edgecolor= "black")
plt.title("Distribución tasa de natalidad")
plt.xlabel("Tasa Natalidad")
plt.ylabel("Total registros")
plt.show()
```

2. ¿Existe una relación aparente entre la tasa de mortalidad infantil y la esperanza de vida? Proporciona una breve descripción de los patrones que observes.


```python
sns.regplot(x = "infant_mortality", y = "life_expectancy", data = df, marker = "v", color = "green")
plt.xticks(rotation = 90)
plt.title("Relación tasa mortalidad infantil y esperanza de vida")
plt.xlabel("Tasa mortalidad infantil")
plt.ylabel("Esperanza de vida")
plt.show()
```

3. ¿Cuál es la mediana de la esperanza de vida en Europa? Proporciona un rango intercuartil aproximado y verifica si existen valores atípicos.


```python
df_europa = df[df["continent"]== "Europe"]
df_europa.sample(10)
```

```python
sns.boxplot(y = "life_expectancy",data = df_europa, width = 0.5, color = "Turquoise")
plt.title("Mediana esperanza de vida Europa")
plt.show()
```

4. ¿Existen relaciones visibles entre la tasa de natalidad y las siguientes variables: tasa de mortalidad infantil, esperanza de vida, densidad de población y población urbana? Usa un subplot. Describe cualquier patrón que encuentres.


```python
plt.figure(figsize = (10,8)) # crear la cuadrícula.
plt.subplot(2,2,1) #filas, columnas, tamaño.
plt.scatter(x = "fertility_rate", y = "infant_mortality", data = df, color = "red")
plt.title("Natalidad vs mortaldiad infantil")
plt.xlabel("Tasa natalidad")
plt.ylabel("tasa mortalidad infantil")

plt.subplot(2,2,2) #filas, columnas, tamaño.
plt.scatter(x = "fertility_rate", y = "life_expectancy", data = df, color = "blue")
plt.title("Natalidad vs esperanza de vida")
plt.xlabel("Tasa natalidad")
plt.ylabel("Esperanza vida")

plt.subplot(2,2,3) #filas, columnas, tamaño.
plt.scatter(x = "fertility_rate", y = "population", data = df, color = "Turquoise")
plt.title("Natalidad vs densidad población")
plt.xlabel("Tasa natalidad")
plt.ylabel("Densidad de población")

plt.subplot(2,2,4) #filas, columnas, tamaño.
plt.scatter(x = "fertility_rate", y = "urban_population", data = df, color = "violet")
plt.title("Natalidad vs población urbana")
plt.xlabel("Tasa natalidad")
plt.ylabel("Población urbana")

plt.tight_layout()
plt.show()
```
