#### 4. Comparando grupos con medias y medianas: países más y menos ricos

Variable base:

* `gdp`

1. Calcula la **mediana** de `gdp` en el conjunto de datos.


```python
mediana_gdp = round(df['gdp'].median())
```

2. Crea dos grupos de países:
   * `grupo_bajo_gdp`: países con `gdp` por debajo (o igual) de la mediana.
   * `grupo_alto_gdp`: países con `gdp` por encima de la mediana.


```python
mascara_bajo_gdp = df['gdp'].le(mediana_gdp)
mascara_alto_gdp = df['gdp'].gt(mediana_gdp)

grupo_bajo_gdp = df[mascara_bajo_gdp]
grupo_alto_gdp = df[mascara_alto_gdp]
```

3. Para cada grupo, calcula la **media y la mediana** de:
   * `life_expectancy`
   * `infant_mortality`


```python
grupo_bajo_gdp['life_expectancy'].mean()
grupo_bajo_gdp['life_expectancy'].median()
grupo_bajo_gdp['infant_mortality'].mean()
grupo_bajo_gdp['infant_mortality'].median()

grupo_alto_gdp['life_expectancy'].mean()
grupo_alto_gdp['life_expectancy'].median()
grupo_alto_gdp['infant_mortality'].mean()
grupo_alto_gdp['infant_mortality'].median()


print(f"La media de esperanza de vida del grupo con bajo PIB es: {round(grupo_bajo_gdp['life_expectancy'].mean()),2}")
print(f"La mediana de esperanza de vida del grupo con bajo PIB es: {round(grupo_bajo_gdp['life_expectancy'].median()),2}")
print(f"La media de mortalidad infantil del grupo con bajo PIB es: {round(grupo_bajo_gdp['infant_mortality'].mean()),2}")
print(f"La median de mortalidad infantil del grupo con bajo PIB es: {round(grupo_bajo_gdp['infant_mortality'].median()),2}")

print("----------------------------------------------------------------------------------------------------------------------------")

print(f"La media de esperanza de vida del grupo con alto PIB es: {round(grupo_alto_gdp['life_expectancy'].mean()),2}")
print(f"La mediana de esperanza de vida del grupo con alto PIB es: {round(grupo_alto_gdp['life_expectancy'].median()),2}")
print(f"La media de mortalidad infantil del grupo con alto PIB es: {round(grupo_alto_gdp['infant_mortality'].mean()),2}")
print(f"La median de mortalidad infantil del grupo con alto PIB es: {round(grupo_alto_gdp['infant_mortality'].median()),2}")
```
