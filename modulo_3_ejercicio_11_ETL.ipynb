```python
df_clientes.columns = df_clientes.columns.str.strip().str.lower()
```

```python
df_clientes.columns
```

```python
columnas_texto = ["first_name","last_name", "city", "country", "gender", "address"]

for col in columnas_texto:
    df_clientes[col] = df_clientes[col].str.strip().str.title()
```

```python
df_clientes.head()
```

```python
df_clientes.info()
```

```python
df_ventas.columns = df_ventas.columns.str.strip().str.lower()
df_ventas.head()
```

```python
df_ventas.info()
```

```python
df_ventas["fecha_venta"] = pd.to_datetime(df_ventas["fecha_venta"])
```

```python
df_ventas.info()
```

```python
df_merge = pd.merge(df_clientes, df_ventas, how = "outer", left_on = "id", right_on = "id_cliente")

df_merge.sample(10)
```
